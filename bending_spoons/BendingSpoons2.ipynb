```python

with open('selected_strings.txt', 'r') as f:
    text = f.read()

```

```python
def weight_of_string(s):
    n = 1
    for l in s:
        n*=1+ord(l)-97
    return n
```

```python
max_str = ""
max_weight = 0
text_list = text.split("\n")
for line in text_list:
    #print(line)
    if len(line) == 10:
        if weight_of_string(line) > max_weight:
            max_weight = weight_of_string(line)
            max_str = line



```

```python
n = 0
for line in text_list:
    #print(line)
    if len(line) == 1:
        n+=1   
```

```python
def shagrams(w, k):
    new_chr = ''
    for c in w:
        new_chr += chr(((ord(c)-97 + k)%26 + 97) )

    return new_chr    

    
```

```python
w_sa = shagrams("while", -4)
w_w = weight_of_string(w_sa)
print(w_w)
```

```python
n = 0
for line in text_list:
    #print(line)
    if len(line) == 4 and weight_of_string(line) == w_w:
        n+=1   
```
