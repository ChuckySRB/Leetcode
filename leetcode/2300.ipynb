```python
def find_barier(potions, barier):
    # Binary search for the leftmost index where potions[idx] >= barier
    left, right = 0, len(potions)
    while left < right:
        mid = (left + right) // 2
        if potions[mid] >= barier:
            left = mid + 1
        else:
            right = mid
    return left
```
